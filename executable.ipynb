# Importing Data

```python
# Module Requests needed for API
import requests as rq

# Extract list of Indicators from INE
series_cd = {
    "0000861",
    "0000862",
    "0000863",
    "0000864",
    "0000865",
    "0000866",
    "0000867",
    "0000868",
    "0000869",
}

# Will Store Variable Info
labels = {
    "cd": [],
    "name": [],
    "meta": [],
}

raw_data = []

for i in series_cd:
    # Get All Information on Data
    url = f"https://www.ine.pt/ine/json_indicador/pindicaMeta.jsp?varcd={i}&lang=PT"
    meta = rq.get(url).json()[0]

    # List Relevant Data Information
    labels["cd"].append(i)
    labels["name"].append(meta["IndicadorNome"])
    labels["meta"].append(url)

    # With every category listed, get category values
    dims = {"dim_num": [], "name": [], "cats": []}

    for dim in meta["Dimensoes"]["Descricao_Dim"][1:]:
        dims["dim_num"].append(dim["dim_num"])
        dims["name"].append(dim["abrv"])
        dims["cats"].append({"cod": [], "value": []})

        for cat in meta["Dimensoes"]["Categoria_Dim"]:
            display(meta)
            keys = list(cat.keys())
            for key in keys:
                if cat[key]["dim_num"] == dims["dim_num"]:
                    dims["cats"]["cod"].append(cat[0]["categ_cod"])
                    dims["cats"]["cod"].append(cat[0]["categ_dsg"])
display(dims)
```
